```python

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

# Copy data sedia ada
df_trap = df.copy()

# Pastikan data depth tersusun
df_trap = df_trap.sort_values("DEPTH").reset_index(drop=True)

# Threshold untuk mengenal pasti sand interval
sand_threshold = 0.70

# Identify potential sand intervals
df_trap["SAND_FLAG"] = (
    df_trap["SAND_PROB"] >= sand_threshold
)

# Identify continuous intervals
df_trap["ZONE"] = (
    df_trap["SAND_FLAG"] !=
    df_trap["SAND_FLAG"].shift()
).cumsum()

# Extract candidate zones
candidates = []

for zone, group in df_trap.groupby("ZONE"):

    if group["SAND_FLAG"].iloc[0]:

        top_depth = group["DEPTH"].min()
        base_depth = group["DEPTH"].max()

        thickness = base_depth - top_depth

        mean_probability = group["SAND_PROB"].mean()

        max_probability = group["SAND_PROB"].max()

        candidates.append({
            "TOP_DEPTH": top_depth,
            "BASE_DEPTH": base_depth,
            "THICKNESS": thickness,
            "MEAN_SAND_PROB": mean_probability,
            "MAX_SAND_PROB": max_probability
        })

# Create candidate table
trap_candidates = pd.DataFrame(candidates)

# Rank candidates
trap_candidates = trap_candidates.sort_values(
    "MEAN_SAND_PROB",
    ascending=False
)

print("AUTOMATED CANDIDATE INTERVALS")
print(trap_candidates)
```

```python

plt.figure(figsize=(10, 8))

plt.plot(
    df_trap["SAND_PROB"],
    df_trap["DEPTH"],
    color="orange",
    label="Sand Probability"
)

plt.axvline(
    0.70,
    color="red",
    linestyle="--",
    label="Sand Threshold"
)

plt.gca().invert_yaxis()

plt.xlabel("Sand Probability")
plt.ylabel("Depth")
plt.title("Automated Candidate Reservoir Intervals")

plt.legend()
plt.grid()

plt.show()
```

```python
import segyio
import numpy as np
import os

# Folder inversion
inv_path = "/data/epic20/APGCE/input_data/challenge1_challenge2/inversion"

# Load SEG-Y files
def load_segy(filepath):
    with segyio.open(filepath, "r", ignore_geometry=True) as f:
        data = np.stack([f.trace[i] for i in range(f.tracecount)])
    return data

# Load S-Impedance
s_imp = load_segy(os.path.join(
    inv_path,
    "Challenge1_Challenge2_Inversion_S_Impedance_cropped.sgy"
))

# Load P-Impedance
p_imp = load_segy(os.path.join(
    inv_path,
    "Challenge1_Challenge2_Inversion_P_Impedance_cropped.sgy"
))

# Load Vp/Vs
vpvs = load_segy(os.path.join(
    inv_path,
    "Challenge1_Challenge2_Inversion_VpVs_cropped.sgy"
))

# Load Total Sand Probability
total_sand = load_segy(os.path.join(
    inv_path,
    "Challenge1_Challenge2_Inversion_TotalSandProbability_cropped.sgy"
))

# Check dimensions
print("S-Impedance:", s_imp.shape)
print("P-Impedance:", p_imp.shape)
print("Vp/Vs:", vpvs.shape)
print("Total Sand:", total_sand.shape)
```

```python
# Check data statistics
# Kita sample data untuk elakkan penggunaan RAM terlalu tinggi

sample_step = 1000

data_sample = {
    "S-Impedance": s_imp[::sample_step],
    "P-Impedance": p_imp[::sample_step],
    "Vp/Vs": vpvs[::sample_step],
    "Sand Probability": total_sand[::sample_step]
}

for name, data in data_sample.items():
    print(f"\n{name}")
    print("Minimum:", np.nanmin(data))
    print("Maximum:", np.nanmax(data))
    print("Mean:", np.nanmean(data))
    print("Median:", np.nanmedian(data))
```
